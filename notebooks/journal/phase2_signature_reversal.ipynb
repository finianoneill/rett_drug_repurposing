## 2. Reversing perturbagens

Top reversers by z-sum, resolved to approved drugs. Which have a plausible Rett mechanism (HDAC inhibition, anticonvulsant, neuroprotective)? Notes:

```python
revs = json.loads(Path('../../data/raw/lincs/reversers.json').read_text())['reversers']
best = {}
for r in revs:
    n = r.get('pert_name')
    if n and (n not in best or r['z_sum'] < best[n]):
        best[n] = r['z_sum']
for n, z in sorted(best.items(), key=lambda x: x[1])[:15]:
    print(f'{z:7.2f}  {n}')
```
